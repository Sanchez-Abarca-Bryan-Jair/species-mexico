```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_observaciones = pd.read_csv('data/Observaciones.csv', sep='\t', low_memory=False)
```

```python
cols_to_drop = [
    'gbifID',
    'datasetKey',
    'occurrenceID',
    'verbatimScientificNameAuthorship',
    'countryCode',
    'locality',
    'individualCount',
    'coordinatePrecision',
    'elevation',
    'elevationAccuracy',
    'depth',
    'depthAccuracy',
    'recordNumber',
    'typeStatus',
    'establishmentMeans'

]

df_observaciones = df_observaciones.drop(columns=cols_to_drop)
```

```python
df_observaciones.info()
```

```python
cols_utiles = [
    "catalogNumber", "class", "order", "family", "genus", "species",
    "decimalLatitude", "decimalLongitude", "eventDate", "year", "month", "day",
    "coordinateUncertaintyInMeters"
]
df_observaciones = df_observaciones[cols_utiles].copy()

# Eliminar filas sin género (no sirven ni para importancia médica ni para especie)
df_observaciones = df_observaciones.dropna(subset=["genus"])

# Rellenar especie faltante con "Genero sp." para no perder registros de importancia médica
df_observaciones["species"] = df_observaciones["species"].fillna(df_observaciones["genus"] + " sp.")

# Calcular el Día del Año (DOY) para fenología
fechas_dt = pd.to_datetime(df_observaciones["eventDate"], errors="coerce")

df_observaciones["fecha"] = fechas_dt.dt.date
df_observaciones["doy"] = fechas_dt.dt.dayofyear
```

```python
fechas_dt = pd.to_datetime(df_observaciones[["year", "month", "day"]])

df_observaciones["fecha"] = fechas_dt.dt.date
df_observaciones["doy"] = fechas_dt.dt.dayofyear.astype("int16")
```

```python
df_observaciones.info()
```

```python
df_observaciones = df_observaciones.drop_duplicates(subset=["catalogNumber"])

for col in ["class", "order", "family", "genus", "species"]:
    df_observaciones[col] = df_observaciones[col].str.strip()

df_observaciones["order"] = df_observaciones["order"].fillna("Sin clasificar")
df_observaciones["family"] = df_observaciones["family"].fillna("Sin clasificar")
```

```python
# Conservar nulos O registros con incertidumbre <= 10,000 metros
df_observaciones = df_observaciones[
    df_observaciones["coordinateUncertaintyInMeters"].isna() | 
    (df_observaciones["coordinateUncertaintyInMeters"] <= 10000)
]
```

```python
import geopandas as gpd

# 1. Convertir observaciones a GeoDataFrame
obs_gdf = gpd.GeoDataFrame(
    df_observaciones,
    geometry=gpd.points_from_xy(
        df_observaciones["decimalLongitude"], 
        df_observaciones["decimalLatitude"]
    ),
    crs="EPSG:4326"
)

# 2. Cargar municipios y estados (INEGI)
mun_gdf = gpd.read_file("data/poligonos/00mun.shp").to_crs(epsg=4326)
ent_gdf = gpd.read_file("data/poligonos/00ent.shp").to_crs(epsg=4326)

# Extraer catálogos limpios para subir después a Neon
df_estados = (
    ent_gdf[["CVE_ENT", "NOMGEO"]]
    .drop_duplicates()
    .rename(columns={"CVE_ENT": "cve_ent", "NOMGEO": "nombre"})
)
df_municipios = (
    mun_gdf[["CVEGEO", "CVE_ENT", "NOMGEO"]]
    .drop_duplicates()
    .rename(columns={"CVEGEO": "cvegeo", "CVE_ENT": "cve_ent", "NOMGEO": "nombre"})
)

# 3. Primer Spatial Join: asignar estado y municipio (descarta puntos fuera de México)
obs_espacial = gpd.sjoin(
    obs_gdf, 
    mun_gdf[["CVEGEO", "CVE_ENT", "geometry"]], 
    how="inner", 
    predicate="within"
).drop(columns=["index_right"])

# 4. Cargar y unir ambas capas de Áreas Naturales Protegidas (CONABIO)
anp_fed = gpd.read_file("data/poligonos/areas_naturales/areas_federales/anpjun2026gw.shp").to_crs(epsg=4326)[["geometry"]]
anp_est = gpd.read_file("data/poligonos/areas_naturales/areas_municipales/anpest25gw.shp").to_crs(epsg=4326)[["geometry"]]

anp_total = pd.concat([anp_fed, anp_est], ignore_index=True)

# 5. Segundo Spatial Join: detectar si cae dentro de cualquier ANP
obs_espacial = gpd.sjoin(
    obs_espacial, 
    anp_total, 
    how="left", 
    predicate="within"
)

# 6. Bandera booleana y eliminación de duplicados por polígonos traslapados
obs_espacial["en_anp"] = obs_espacial["index_right"].notna()
obs_espacial = obs_espacial.drop_duplicates(subset=["catalogNumber"])

# 7. Regresar a DataFrame normal y renombrar columnas
df_observaciones = pd.DataFrame(
    obs_espacial.drop(columns=["geometry", "index_right"])
).rename(columns={
    "CVE_ENT": "cve_ent",
    "CVEGEO": "cvegeo"
})

print(f"Registros dentro de México: {len(df_observaciones):,}")
print(f"Registros dentro de ANP: {df_observaciones['en_anp'].sum():,}")
```
